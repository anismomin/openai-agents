```python
from agents import Agent, Runner, AsyncOpenAI, set_default_openai_client, set_default_openai_api, set_tracing_disabled, OpenAIChatCompletionsModel, function_tool, enable_verbose_stdout_logging
from google.colab import userdata

gemini_api_key=userdata.get('GEMINI_API_KEY')

# Disabled tracing as we are working non openai model
set_tracing_disabled(True)

# Check tracing in logs
enable_verbose_stdout_logging()
set_default_openai_api('chat_completions')

external_client = AsyncOpenAI(
    api_key=gemini_api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

set_default_openai_client(external_client)

model = OpenAIChatCompletionsModel(
    model="gemini-2.0-flash",
    openai_client=external_client
)

@function_tool
def devOps(task: str) -> str:
    '''Devops tool to handle cloud native deployment make scalable solutions

    Args:
        task: Any task related to deployment

    Returns:
        Complete deployment task and return status.
    '''

    print(f"Devops tool called")

    return f"Deployment task is compete {task}"


@function_tool
def openAiAgent(task: str) -> str:
    '''Agentic AI export specialist on openAi

    Args:
        task: Any task agent development and design workflow

    Returns:
        Create AI agents using industies best practices.
    '''

    print(f"OpenAiAgent tool called")

    return f"Completed AI agents related task {task}"


webdev_agent : Agent = Agent(
    name="WebDev Agent",
    instructions="Web Developer expert on javascript and python language",
    model="gemini-1.5-flash",
    handoff_description="Web development expert"
)

mobiledev_agent : Agent = Agent(
    name="MobileDev Agent",
    instructions="Mobile app development expert",
    model="gemini-2.0-flash",
    handoff_description="Mobile app development expert"
)

triage_agent : Agent = Agent(
    name="Panacloud",
    instructions="IT Company Specialized on cutting edge technologies",
    model="gemini-2.0-flash",
    tools=[devOps, openAiAgent],
    handoffs=[webdev_agent, mobiledev_agent]
)

response = Runner.run_sync(triage_agent, 'I want to build AI agent and develop a website and web to deploy on cloud and it should be scalable')

print(response.final_output)

```
